```python
import logging

from llama_index.core import Settings
from llama_index.llms.ollama import Ollama
from llama_index.embeddings.huggingface import HuggingFaceEmbedding

from src.tools.rag_pipeline import RAGTool
from src.utils.logger_config import logger
```

```python
fl = logging.getLogger("filelock")
fl.disabled = True
```

```python
# --- 1. Configure the LLM ---
# The RAGTool's chat engine needs an LLM to synthesise the final answer.
# This tells LlamaIndex to use the local 'phi3:mini' model.
logger.info("Configuring LLM and Embedding Model...")

# Configure the LLM to use the local 'phi3:3.8b-mini-4k-instruct-q4_0' model
Settings.llm = Ollama(model="phi3:3.8b-mini-4k-instruct-q4_0", request_timeout=300.0)

# Configure the embedding model to use the same local model from the ingestion pipeline
Settings.embed_model = HuggingFaceEmbedding(model_name="all-MiniLM-L6-v2")

logger.info("LLM and Embedding Model configured successfully.")
```

```python
# --- 2. Initialise and Test the RAGTool ---
# This will load the index and set up the chat engine automatically.
try:
    rag_tool = RAGTool()

    # --- 3. Run a Test Query ---
    question = "Based on their latest 10-K filing, what are the main risks for Microsoft?"
    response = rag_tool.query(question)
    
    print("\n" + "="*50)
    print(f"Question: {question}")
    print("\n" + "-"*50 + "\n")
    print(f"Response:\n{response}")
    print("="*50)

except Exception as e:
    logger.error(f"An error occurred during the RAG tool test: {e}", exc_info=True)
```
