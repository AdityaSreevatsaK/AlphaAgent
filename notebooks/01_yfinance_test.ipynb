# `01_yfinance_test`
## What This Code Does
- Imports `yfinance`: The essential library for fetching the data.
- Selects Tickers: It uses a small list of tickers (AAPL, NVDA, TSLA) for quick testing, based on the companies you selected in your rationale document.
- Fetches Company Info: The yf.Ticker("AAPL").info command retrieves a comprehensive dictionary of information for Apple.
- Validates Key Metrics: It shows how to safely extract specific, important data points like price, P/E ratio, and market cap. Using .get() is a safe way to access dictionary keys that may not always be present.
- Fetches Historical Data: It demonstrates how to get daily performance data (Open, High, Low, Close, Volume) for a given period.

```python
import yfinance as yf
import pandas as pd
```

```python
pd.set_option("display.max_columns", None)
```

```python
# --- 1. Select Tickers ---
# Using a few tickers from my initial list for testing
tickers_to_test = ["AAPL", "NVDA", "TSLA"]
```

```python
# 2. Fetching company information.
print("Detailed Company Information")
aapl_ticker = yf.Ticker("AAPL")
aapl_info = aapl_ticker.info
print(f"Successfully fetched {len(aapl_info)} data points for 'AAPL'.")
```

```python
# Validate key data points
print("Validating key data points for 'AAPL'.")
current_price = aapl_info.get("currentPrice", "N/A")
pe_ratio = aapl_info.get("trailingPE", "N/A")
market_cap = aapl_info.get("marketCap", "N/A")
enterprise_value = aapl_info.get("enterpriseValue", "N/A")
fifty_two_week_high = aapl_info.get("fiftyTwoWeekHigh", "N/A")
dividend_yield = aapl_info.get("dividendYield", "N/A")

print(f"Current Price: {current_price}")
print(f"P/E Ratio (Trailing): {pe_ratio}")
print(f"Market Cap: {market_cap:,}")  # Using :, to format with commas
print(f"Enterprise Value: {enterprise_value:,}")
print(f"52-Week High: {fifty_two_week_high}")
print(
    f"Dividend Yield: {dividend_yield * 100:.2f}%"
    if isinstance(dividend_yield, float)
    else "N/A"
)
```

```python
# 5. Fetch historical data.
print("Fetching historical data for 'NVDA'.")
nvda_ticker = yf.Ticker("NVDA")
nvda_historical_data = nvda_ticker.history(period="1mo")
print("Successfully fetched historical data for 'NVDA'. Showing the last 5 days:")
display(nvda_historical_data.tail())
```
